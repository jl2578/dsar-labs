## 3. Lists and dictionaries as ways to organize related information

A list stores values in order. A dictionary stores labeled values. A Python dictionary is not the same thing as an official research data dictionary, such as the scientific definitions of a variable in a biomedical dataset.

Here, the goal is intuition: the list keeps related values together, and the dictionary attaches simple labels to those values.

### A quick note about study records

In this course, a **participant** is one person enrolled in a study. A **session** is one occasion when information is collected from that participant, such as a baseline visit or a later follow-up visit.

A **participant-session record** contains information about one participant at one session. For example, a record might include a participant ID, a session ID, the number of reported cannabis use days, and whether the participant reported any cannabis use.

The examples below are fictional and are provided for practice only. They demonstrate how Python can organize information that could later appear in a research dataset.

````python
# Demo: Organizing one participant-session record with a list and a dictionary

session_values = [
    "sub-DSARM0001",
    "ses-01",
    3,
    True
]

session_record = {
    "participant_id": "sub-DSARM0001",
    "session_id": "ses-01",
    "use_days": 3,
    "reported_use": True
}

print("List version:")
print(session_values)
print("List type:", type(session_values))

print("Dictionary version:")
print(session_record)
print("Dictionary type:", type(session_record))
````

### What to notice

- The list contains values, but the meaning depends on position.
- The dictionary attaches labels to values, which makes the record easier to read.
- The dictionary is only a simple Python container here, not an official data dictionary.

### ABCD variable names as structured addresses

ABCD variable names function like structured addresses. A name can help identify where a variable belongs in the data system, but it does not provide enough information to interpret the variable scientifically.

A basic ABCD variable name follows this structure:

`domain_source_table_item`

- Single underscores separate the four main components.
- The **domain** identifies the broad content area.
- The **source** identifies the respondent or other source/recipient of the data.
- The **table** identifies the instrument, form, or related group of variables.
- A three-digit ending such as `004` generally identifies an item.
- A text ending such as `prsum` can provide a clue that a variable is a summary score.
- Double underscores can mark nested groups, subitems, versions, or tags in more complex names.

| Variable | Domain | Source | Table or group | Item or score clue |
|---|---|---|---|---|
| `su_y_mjprob_004` | `su`: substance use | `y`: youth | `mjprob` | `004`: item |
| `su_y_mjprob_prsum` | `su`: substance use | `y`: youth | `mjprob` | `prsum`: summary-score clue |
| `su_y_sui__use__mj__puff_001__l` | `su`: substance use | `y`: youth | `sui` with nested topic components | `001`: item; `__l`: longitudinal tag |

The name can help you identify the domain, source, table, and item or score structure. It cannot establish the exact item wording, response options, reference period, coding, documented metadata, scoring rule, or scientific meaning. Treat components such as `004` and `prsum` as structural clues, not as sufficient evidence for interpretation.

A **pandas dtype** describes how values are currently represented in a DataFrame. It is not the same as the study's documented `type_data`, the variable's `type_var` or scientific role, or its `type_level`. Neither a variable name nor a pandas dtype is sufficient to determine what a variable scientifically represents. Exercise 2 will formally introduce these distinctions.

**Source and further reading:** [ABCD Study data documentation: Naming convention](https://docs.abcdstudy.org/latest/documentation/curation/naming.html).

## 6. Identifiers, DataFrame index, and row units

A DataFrame has rows and columns. Now we ask what one row represents. That is the row unit. Identifiers help us investigate row units.

`participant_id` and `session_id` are study identifiers. The pandas index is a row label created by pandas and is not automatically a study identifier.

````python
# Demo: The pandas index is not the same thing as a study identifier

display(example_dataframe)

print("Pandas index:")
print(example_dataframe.index)

print("\nParticipant IDs:")
print(example_dataframe["participant_id"])
````

### What to notice

The DataFrame has a pandas index on the left. Those index values are row labels created by pandas.

The participant IDs are stored in a column. They are study identifiers, not pandas index values.

This distinction matters because researchers should use documented identifiers, not row numbers, to understand what records represent.

````python
# Demo: Use identifiers to investigate the row unit of su_y_sui

sui_row_count = su_y_sui.shape[0]
sui_participant_count = su_y_sui["participant_id"].nunique()
sui_participant_session_count = su_y_sui[["participant_id", "session_id"]].drop_duplicates().shape[0]

sui_participant_id_repeats = su_y_sui["participant_id"].duplicated().any()
sui_participant_session_repeats = su_y_sui[["participant_id", "session_id"]].duplicated().any()

identifier_check = pd.DataFrame({
    "check": [
        "number of rows",
        "unique participant_id values",
        "unique participant_id + session_id pairs",
        "does participant_id repeat?",
        "does participant_id + session_id repeat?"
    ],
    "result": [
        sui_row_count,
        sui_participant_count,
        sui_participant_session_count,
        sui_participant_id_repeats,
        sui_participant_session_repeats
    ]
})

display(identifier_check)
````

### What to notice

These checks help us reason about the row unit.

If `participant_id` repeats, then the table probably has more than one row per participant.

If each `participant_id` + `session_id` pair is unique, that suggests each row may represent one participant at one session.

This is evidence from the table structure. We should still confirm row-unit meaning with documentation before making analytic claims.

## 7. Simple counts for orientation

Section 5 inspected table structure. Section 6 investigated row units and identifiers. Now we look inside one column to see what values appear.

A simple count is a first orientation step. It tells us what appears in a column before we interpret it.

````python
print("Counts for reported_use in example_dataframe:")
print(example_dataframe["reported_use"].value_counts(dropna=False))
````

### What to notice

The column `reported_use` contains values such as `True`, `False`, and missing.

The argument `dropna=False` tells pandas to include missing values in the count.

This is useful because missing values are part of the data structure. We should not ignore them automatically.

````python
sui_use_counts = su_y_sui["su_y_sui__use__mj__puff_001__l"].value_counts(dropna=False)

print("Counts for su_y_sui__use__mj__puff_001__l:")
print(sui_use_counts)
````

### What to notice

`dropna=False` includes missing values. Counts show what values appear, but counts do not explain what the values mean. In real ABCD-style analysis, researchers should cross-reference official documentation to confirm variable meaning, response options, branching logic, missingness, and version information. In this notebook, counts are used only for orientation.
